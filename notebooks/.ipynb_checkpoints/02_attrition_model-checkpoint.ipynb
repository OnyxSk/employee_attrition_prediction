```python
import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    RocCurveDisplay
)

import matplotlib.pyplot as plt
```

```python
from pathlib import Path
import os

print("📍 Current working directory:", os.getcwd())

df = pd.read_csv("../data/raw/Employee.csv")
df.head()
```

```python
df.shape, df.columns
```

```python
df["LeaveOrNot"].value_counts(normalize=True)
```

```python
TARGET = "LeaveOrNot"

X = df.drop(columns=[TARGET])
y = df[TARGET]

X.shape, y.shape
```

```python
cat_cols = X.select_dtypes(include=["object"]).columns.tolist()
num_cols = X.select_dtypes(exclude=["object"]).columns.tolist()

cat_cols, num_cols
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

X_train.shape, X_test.shape
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), num_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
    ]
)

model = LogisticRegression(max_iter=2000)

pipeline = Pipeline(steps=[
    ("preprocessing", preprocessor),
    ("model", model)
])

pipeline
```

```python
pipeline.fit(X_train, y_train)

y_pred = pipeline.predict(X_test)

acc = accuracy_score(y_test, y_pred)
print("Accuracy:", acc)
```

```python
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
print(classification_report(y_test, y_pred, target_names=["No Leave", "Leave"]))
```

```python
y_proba = pipeline.predict_proba(X_test)[:, 1]
auc = roc_auc_score(y_test, y_proba)
print("ROC AUC:", auc)
```

```python
RocCurveDisplay.from_predictions(y_test, y_proba)
plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    random_state=42,
    class_weight="balanced"  # très important en churn
)

rf_pipeline = Pipeline(steps=[
    ("preprocessing", preprocessor),
    ("model", rf_model)
])
```

```python
rf_pipeline.fit(X_train, y_train)

rf_pred = rf_pipeline.predict(X_test)
rf_proba = rf_pipeline.predict_proba(X_test)[:, 1]

print("Accuracy:", accuracy_score(y_test, rf_pred))
print("ROC AUC:", roc_auc_score(y_test, rf_proba))
print(classification_report(y_test, rf_pred, target_names=["No Leave", "Leave"]))
```

```python
RocCurveDisplay.from_predictions(y_test, y_proba, name="Logistic Regression")
RocCurveDisplay.from_predictions(y_test, rf_proba, name="Random Forest")
plt.show()
```

## Model comparison

Two models were evaluated:
- Logistic Regression (baseline, interpretable)
- RandomForest (non-linear, more expressive)

RandomForest shows improved performance in terms of ROC AUC, indicating a better ability to rank employees by churn risk.
However, Logistic Regression remains useful as a simple and interpretable baseline.

```python
# Récupérer les noms des features après preprocessing
feature_names = (
    rf_pipeline.named_steps["preprocessing"]
    .get_feature_names_out()
)

importances = rf_pipeline.named_steps["model"].feature_importances_

feat_imp = (
    pd.DataFrame({
        "feature": feature_names,
        "importance": importances
    })
    .sort_values("importance", ascending=False)
    .head(15)
)

feat_imp
```

```python
feat_imp.plot(
    x="feature",
    y="importance",
    kind="barh",
    figsize=(8, 6)
)

plt.title("Top 15 Feature Importances (RandomForest)")
plt.xlabel("Importance")
plt.ylabel("")
plt.gca().invert_yaxis()
plt.show()
```

## Final conclusions

This project combined exploratory analysis and predictive modeling to study employee attrition.

Key takeaways:
- Attrition is strongly associated with compensation level and experience.
- Mid-level employees (PaymentTier 2) represent a particularly high-risk group.
- RandomForest improves churn prediction compared to a linear baseline.

From a business perspective, retention strategies should focus on:
- Compensation review for mid-tier employees
- Career development and engagement for early-career profiles
- Monitoring employees with benching history
